# 데이터 로드하기


```python
import pandas as pd


class DataPreprocessor:
    def __init__(self, file_path):
        self.file_path = file_path
        self.df = None

    # 1. 데이터 불러오기
    def load_data(self):
        self.df = pd.read_excel(self.file_path)

        print("=" * 50)
        print("Data Load Complete")
        print(f"Rows    : {self.df.shape[0]}")
        print(f"Columns : {self.df.shape[1]}")
        print("=" * 50)

        return self.df

    # 2. 기본 정보 확인
    def data_info(self):
        print("\n========== DATA INFO ==========")
        print(self.df.info())

        print("\n========== MISSING VALUES ==========")
        print(self.df.isnull().sum())

    # 3. DateTime 변환
    def convert_datetime(self):
        self.df["DateTime"] = pd.to_datetime(self.df["DateTime"])

    # 4. 시간순 정렬
    def sort_by_time(self):
        self.df = self.df.sort_values("DateTime").reset_index(drop=True)

    # 5. Machine + Product + Recipe 기준 그룹 생성
    def make_groups(self):

        groups = {}

        grouped = self.df.groupby(
            ["Machine_ID", "Product_ID", "Recipe_ID"]
        )

        for key, group in grouped:
            group = group.sort_values("DateTime").reset_index(drop=True)
            groups[key] = group

        print(f"\nTotal Groups : {len(groups)}")

        return groups

    # 6. 그룹 정보 출력
    def print_group_summary(self, groups):

        print("\n========== GROUP SUMMARY ==========")

        for key, group in groups.items():
            print(
                f"{key}  ->  {len(group)} rows"
            )
```

```python
from preprocessing import DataPreprocessor


FILE_PATH = "data/suni dataset.xlsx"

processor = DataPreprocessor(FILE_PATH)

processor.load_data()

processor.data_info()

processor.convert_datetime()

processor.sort_by_time()

groups = processor.make_groups()

processor.print_group_summary(groups)
```
